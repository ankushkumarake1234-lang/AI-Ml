```python
# question no 8
X = np.array([
    [50, 1200, 3],
    [80, 1500, 9],
    [65, 900, 5],
    [95, 2000, 7]
])

col_min = X.min(axis=0)
col_max = X.max(axis=0)

normalized = (X - col_min) / (col_max - col_min)

print(normalized)

print("Column minimum:", normalized.min(axis=0))
print("Column maximum:", normalized.max(axis=0))
```

```python
# question no 9
scores = np.array([94, 61, 77, 45, 90, 88, 59, 72])
medals = np.where(
    scores >= 90,
    "Gold",
    np.where(
        scores >= 75,
        "Silver",
        np.where(
            scores >= 60,
            "Bronze",
            "None"
        )
    )
)
print(medals)
print("Gold:", np.sum(medals == "Gold"))
print("Silver:", np.sum(medals == "Silver"))
print("Bronze:", np.sum(medals == "Bronze"))
print("None:", np.sum(medals == "None"))
```

```python
# question no 14
hours = np.array([2, 8, 5, 10, 3, 7])
attendance = np.array([55, 92, 70, 98, 60, 85])
# 1. Convert to column vectors
hours_col = hours[:, None]
attendance_col = attendance[:, None]

print(hours_col.shape)
print(attendance_col.shape)

# 2. Two-column table
table = np.hstack((hours_col, attendance_col))

print(table)
print(table.shape)

# 3. Total column
total = hours + attendance
total_col = total[:, None]

# 4. Three-column table
three_col = np.hstack((hours_col, attendance_col, total_col))

print(three_col)
print(three_col.shape)

# 5. Weighted score
w = np.array([0.5, 0.2, 0.3])

score = three_col @ w

print(score)
print(score.shape)

# 6. Above-average classification
average = score.mean()

result = np.where(score > average, 1, 0)

print("Average:", average)
print("Result:", result)
print(result.shape)
```

```python
# question no 17
np.random.seed(7)

temps = np.random.randint(15, 41, 28)

weeks = temps.reshape(4, 7)

weekly_avg = weeks.mean(axis=1)

hottest_week = np.argmax(weekly_avg)

days_above_35 = np.sum(temps > 35)

print("Temperatures:")
print(temps)

print("Weeks:")
print(weeks)

print("Weekly averages:")
print(weekly_avg)

print("Hottest week index:", hottest_week)
print("Number of days above 35:", days_above_35)
```

```python
# question no 18
np.random.seed(10)

A = np.random.randint(1, 10, (3, 3))
B = np.random.randint(1, 10, (3, 3))

AB = A @ B
BA = B @ A

print("A:")
print(A)

print("B:")
print(B)

print("A @ B:")
print(AB)

print("B @ A:")
print(BA)

print("Are they equal?", np.array_equal(AB, BA))
```

```python
# question no 19
marks = np.array([
    [45, 28, 67, 80],
    [33, 90, 55, 41],
    [72, 66, 30, 58],
    [88, 79, 91, 84],
    [25, 40, 38, 52]
])

marks = np.maximum(marks, 35)

totals = marks.sum(axis=1)

topper_index = np.argmax(totals)

print("Updated marks:")
print(marks)

print("Totals:", totals)

print("Topper index:", topper_index)
print("Topper total:", totals[topper_index])
```
