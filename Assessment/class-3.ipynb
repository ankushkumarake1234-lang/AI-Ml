```python
import pandas as pd
```

```python
t = pd.read_csv('Titanic-Dataset.csv')
# question no 1
rows, columns = t.shape
print(f"Rows: {rows}, Columns: {columns}")
# Method 2: Using len()
print(f"Rows: {len(t)}, Columns: {len(t.columns)}")
```

```python
# question no 2
print(t.columns.tolist())
```

```python
# question no 3
# First 7 rows
print("First 7 rows:")
print(t.head(7))
# Last 3 rows
print("\nLast 3 rows:")
print(t.tail(3))
```

```python
# question no 4
# Run info to see missing values
t.info()
# More specifically, to see missing values per column:
missing_values = t.isnull().sum()
print("\nMissing values per column:")
print(missing_values[missing_values > 0])
```

```python
# question no 5
# See which columns are included
print(t.describe().columns.tolist())

# Compare with all columns
print("\nAll columns:")
print(t.columns.tolist())
```

```python
# question no 6
# a) How many different values?
print(f"Number of unique values: {t['Pclass'].nunique()}")

# b) What are they?
print(f"Unique values: {t['Pclass'].unique()}")

# c) Count of each value
print(f"Value counts:\n{t['Pclass'].value_counts()}")
```

```python
# question no 7
# Method 1: Using value_counts()
print(t['Sex'].value_counts())

# Method 2: Using conditional filtering
male_count = len(t[t['Sex'] == 'male'])
female_count = len(t[t['Sex'] == 'female'])
print(f"Male: {male_count}, Female: {female_count}")
```

```python
# question no 8
# Get Age statistics
Age_stats = t['Age'].describe()
mean_Age = Age_stats['mean']
median_Age = Age_stats['50%']

print(f"Mean Age: {mean_Age:.2f}")
print(f"Median Age (50%): {median_Age:.2f}")
print(f"Difference: {mean_Age - median_Age:.2f}")

# Which is larger?
if mean_Age > median_Age:
    print("Mean is larger than median")
else:
    print("Median is larger than mean")
```

```python
# question no 9
# Using shape and info
total_rows = t.shape[0]  
non_null_age = t['Age'].count()  
missing_age = total_rows - non_null_age
percentage_missing = (missing_age / total_rows) * 100
print(f"Total rows: {total_rows}")
print(f"Non-null age values: {non_null_age}")
print(f"Missing age values: {missing_age}")
print(f"Percentage missing: {percentage_missing:.2f}%")
# In comment: 19.87% of age values are missing
```
