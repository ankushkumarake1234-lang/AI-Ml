```python
import pandas as pd
```

```python
# Part A – tips.csv
tips = pd.read_csv('tips.csv')
titanic = pd.read_csv('Titanic-Dataset.csv')
```

```python
# question A1
tips['tip_percent'] = (tips['tip'] / tips['total_bill'] * 100).round(1)
print(tips[['total_bill', 'tip', 'tip_percent']].head())
```

```python
# question A2
print(tips.iloc[9, tips.columns.get_loc('total_bill')])
print(tips.iloc[-3:, :2])
```

```python
# question A3
selected = tips.loc[20:25, 'sex':'size']
print(selected)
print("Number of rows:", len(selected))
```

```python
# question A4
not_dinner = tips[tips['time'] != 'Dinner'].shape[0]
print("Not Dinner:", not_dinner)
party_tip = tips[(tips['size'] >= 3) & (tips['tip'] > 4)].shape[0]
print("Party >=3 and tip >4:", party_tip)
```

```python
# question A5
tips['service'] = 'normal'
tips.loc[tips['tip_percent'] > 20, 'service'] = 'generous'
print(tips['service'].value_counts())
```

```python
# question B1
embarked_counts = titanic['Embarked'].value_counts()
print(embarked_counts)
total_embarked = embarked_counts.sum()
print("Sum of the three port counts:", total_embarked)
print("Is it 891?", total_embarked == 891)
```

```python
# question B2
women_third = titanic[(titanic['Sex'] == 'female') & (titanic['Pclass'] == 3)]
count_women_third = women_third.shape[0]
print("Women in third class:", count_women_third)
survived_women_third = women_third[women_third['Survived'] == 1].shape[0]
print("Survived among them:", survived_women_third)
```

```python
# question B3
high_fare = titanic[titanic['Fare'] > 200]
print(high_fare[['Name', 'Age', 'Fare']])
print("Number of passengers with Fare > 200:", high_fare.shape[0])
```

```python
# question B4
first_class = titanic[titanic['Pclass'] == 1]
third_class = titanic[titanic['Pclass'] == 3]
survived_first = first_class[first_class['Survived'] == 1].shape[0]
total_first = first_class.shape[0]
pct_first = survived_first / total_first * 100
survived_third = third_class[third_class['Survived'] == 1].shape[0]
total_third = third_class.shape[0]
pct_third = survived_third / total_third * 100
print(f"First class survival: {pct_first:.2f}%")
print(f"Third class survival: {pct_third:.2f}%")
```

```python
# question B5
titanic['age_group'] = 'adult'
titanic.loc[titanic['Age'] < 18, 'age_group'] = 'child'
print(titanic['age_group'].value_counts())
```

```python
print("Total passengers:", len(titanic))
print("Missing Age:", titanic['Age'].isnull().sum())
print("Children (Age<18):", (titanic['Age'] < 18).sum())
print("Adults (Age>=18):", (titanic['Age'] >= 18).sum())
print("After assignment, 'adult' count includes missing:", titanic[titanic['age_group'] == 'adult'].shape[0])
```
