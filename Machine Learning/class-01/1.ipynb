```python
from sklearn.model_selection import train_test_split
```

```python
x = df.drop(columns=['Survived']) # independent features
y = df['Survived'] # dependent features
print(x.head(1))
print(y.head(1))
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,random_state=42,)
```

```python
cat_cols = ['Sex','Embarked']
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import accuracy_score
```

```python
ct = ColumnTransformer([('cat', OneHotEncoder(sparse_output=False), cat_cols)],remainder='passthrough')
x_train_final = ct.fit_transform(x_train)
x_test_final = ct.transform(x_test)
```

```python
from sklearn.metrics import accuracy_score
from sklearn.neighbors import KNeighborsClassifier
model = KNeighborsClassifier(n_neighbors=5)
model.fit(x_train_final,y_train)
```
