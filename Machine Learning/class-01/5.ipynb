```python
# 4. Cleaning
# blanks per column
df.isnull().sum()
```

```python
# repeated rows
print(df.duplicated().sum())

# remove repeated rows
df_clean = df.drop_duplicates()
```

```python
# remove rows where one column is blank
df_clean = df_clean.dropna(subset=['internships'])

# fill blanks with a value
df_clean['prep_hours'] = df_clean['prep_hours'].fillna(0)

# change the type
df_clean['internships'] = df_clean['internships'].astype(int)
```

```python
# new column from existing columns
df_clean['cgpa_percent'] = (df_clean['cgpa'] * 9.5).round(1)
df_clean[['cgpa', 'cgpa_percent']].head()
```
