```python
# Using .loc[], create a new column fare_level and set it to 'High' where Fare > 30 and 'Low' everywhere else — do this with two separate .loc[] assignment lines.

df.loc[df['Fare'] > 30, 'fare_level'] = 'High'
df.loc[df['Fare'] <= 30, 'fare_level'] = 'Low'
print(df[['Fare', 'fare_level']].head())
```

```python
# Q4.2 Recreate the same idea as a new column fare_level_v2, but this time in a single line using np.where() instead of .loc[].

df['fare_level_v2'] = np.where(df['Fare'] > 30, 'High', 'Low')
print(df[['Fare', 'fare_level_v2']].head())
```

```python
# Q5.1 Filter df for female passengers (Sex == 'female') who survived (Survived == 1) and were traveling in 1st class (Pclass == 1). How many such passengers are there?

df_filtered = df[(df['Sex'] == 'female') & (df['Survived'] == 1) & (df['Pclass'] == 1)]
num_passengers = len(df_filtered)
print(f"Number of female passengers who survived and were in 1st class: {num_passengers}")
```

```python
# Q5.2 Filter df for passengers who are NOT in 3rd class using ~, and whose Age is missing (.isnull()).


df_filtered_not_3rd_class = df[~(df['Pclass'] == 3) & (df['Age'].isnull())]
num_passengers_not_3rd_class = len(df_filtered_not_3rd_class)
print(f"Number of passengers NOT in 3rd class with missing Age: {num_passengers_not_3rd_class}")
```

```python
#  Q6.1 Check how many rows in df have a missing Age, then create df_clean by dropping only those rows using .dropna(subset=['Age']).

df_missing_age = df[df['Age'].isnull()]
num_missing_age = len(df_missing_age)
print(f"Number of rows with missing Age: {num_missing_age}")
```

```python
# Q6.2 Compare the .shape of df before and after plain .dropna() (which drops a row if ANY column has a missing value). What is the difference, and why?


df_shape_before = df.shape
df_clean_dropna = df.dropna()
df_shape_after = df_clean_dropna.shape

print(f"Shape of df before dropping rows: {df_shape_before}")
print(f"Shape of df after dropping rows: {df_shape_after}")

difference = df_shape_before[0] - df_shape_after[0]
print(f"Number of rows dropped: {difference}")
```

```python
# Q7.1 Group df by Pclass and use .first() and .last() to see the first and last passenger listed for each class.

df_grouped = df.groupby('Pclass')
first_passengers = df_grouped.first()
last_passengers = df_grouped.last()
print("First passengers in each class:")   
```

```python
# Q7.2 Group df by Embarked and compare .size() vs .count(). Are the numbers the same for every column? Why might they differ for a column like Age?

df_grouped_embarked = df.groupby('Embarked')
size_per_embarked = df_grouped_embarked.size()
count_per_embarked = df_grouped_embarked.count()
print("Size per Embarked:")
print(size_per_embarked)
print("\nCount per Embarked:")
print(count_per_embarked)
```

```python
# Q7.3 Group df by Pclass and use .size() to count passengers per class, then use .count() on just the Age column to see how many passengers per class have a known age.

df_grouped_pclass = df.groupby('Pclass')
size_per_pclass = df_grouped_pclass.size()
count_age_per_pclass = df_grouped_pclass['Age'].count()
print("Size per Pclass:")
print(size_per_pclass)
print("\nCount of known Age per Pclass:")
print(count_age_per_pclass)

```

```python
# Q8.1 Group df by Pclass and use named aggregation to build a summary with columns avg_fare (mean of Fare) and max_age (max of Age).


df_summary = df.groupby('Pclass').agg(avg_fare=('Fare', 'mean'), max_age=('Age', 'max'))
print(df_summary)
```

```python
# Q8.2 Group df by Pclass and Sex together, and use named aggregation to create avg_age (mean of Age) and total_passengers (count of PassengerId).


df_summary_grouped = df.groupby(['Pclass', 'Sex']).agg(avg_age=('Age', 'mean'), total_passengers=('PassengerId', 'count'))
print(df_summary_grouped)
```

```python
# Q9.1 Group df by Pclass and use .agg() with a lambda function to find the range (max - min) of Fare for each class.

df_fare_range = df.groupby('Pclass').agg(fare_range=('Fare', lambda x: x.max() - x.min()))
print(df_fare_range)
```

```python
# Q9.2 Write your own custom function called age_range(x) that returns x.max() - x.min(). Group df by Embarked and pass this function into .agg() to find the age range for each embarkation port.

def age_range(x):
    return x.max() - x.min()

df_age_range = df.groupby('Embarked').agg(age_range=('Age', age_range))
print(df_age_range)
```

```python
# Q10.1 Create a pivot table from df showing the average Fare, with Pclass as rows and Sex as columns.

pivot_table_avg_fare = df.pivot_table(values='Fare', index='Pclass', columns='Sex', aggfunc='mean') 
print(pivot_table_avg_fare)
```

```python
# Q10.2 Create a pivot table from df showing the survival rate (mean of Survived), with Pclass as rows and Sex as columns.

pivot_table_survival_rate = df.pivot_table(values='Survived', index='Pclass', columns='Sex', aggfunc='mean')
print(pivot_table_survival_rate)
```

```python
# Q11.1 Group df by Embarked and keep only the groups that have more than 100 passengers, using .groupby().filter().

df_filtered = df.groupby('Embarked').filter(lambda x: x.shape[0] > 100)
print(df_filtered)
```

```python
# Q11.2 Group df by Pclass and keep only the groups where the average Fare is greater than 20.

df_filtered = df.groupby('Pclass').filter(lambda x: x['Fare'].mean() > 20)
print(df_filtered)
```

```python
# C1. Using df, build a summary table (via pivot_table or groupby + named agg()) showing, for each Pclass: average Fare, average Age, and total number of passengers — sorted from highest to lowest average Fare.

print(df.groupby('Pclass').agg(avg_fare=('Fare', 'mean'), avg_age=('Age', 'mean'), total_passengers=('PassengerId', 'count')).sort_values(by='avg_fare', ascending=False))
```

```python
# C2. Drop rows with a missing Age from df, create a new column age_group using conditional assignment ('Child' if Age < 18 else 'Adult'), then group by Pclass and age_group to find the survival rate for each combination.

print(df.dropna(subset=['Age']).assign(age_group=lambda x: np.where(x['Age'] < 18, 'Child', 'Adult')).groupby(['Pclass', 'age_group'])['Survived'].mean())


```

```python
# C3. Take a random sample of 200 passengers from df using .sample(n=200, random_state=42). From this sample, filter passengers who are male and did not survive. Then group this filtered data by Pclass, and using named aggregation, create avg_fare and passenger_count.


print(df.sample(n=200, random_state=42).query("Sex == 'male' and Survived == 0").groupby('Pclass').agg(avg_fare=('Fare', 'mean'), passenger_count=('PassengerId', 'count')))        
```
