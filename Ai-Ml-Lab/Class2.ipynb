```python
# Q3.1 Filter and display all cars with mpg greater than 30.
print(df[df['mpg'] > 30])
```

```python
# Q4.1 Sort the dataset by mpg in descending order and show the top 5 most fuel-efficient cars (name + mpg).
print(df.sort_values(by='mpg', ascending=False)[['name', 'mpg']].head(5))
```

```python
# Q4.2 Sort the dataset first by weight (ascending) and then by horsepower (descending).
print(df.sort_values(by=['weight', 'horsepower'], ascending=[True, False]))
```

```python
# Q5.1 Check how many missing values exist in each column of df.
print(df.isnull().sum())
```

```python
# Q5.2 Fill any missing values in the horsepower column with the column's mean value.
print(df['horsepower'].fillna(df['horsepower'].mean(), inplace=True))
```

```python
# Q5.3 Find the total number of missing values in the entire dataset (a single number) using .isnull().sum().sum().
print(df.isnull().sum().sum())
```

```python
# Q6.1 Add a new column weight_kg by converting weight (in lbs) to kilograms (1 lb = 0.4536 kg).
df['weight_kg'] = df['weight'] * 0.4536
print(df[['weight', 'weight_kg']])
```

```python
# Q6.2 Add a new column efficiency using .apply() with a lambda: label a car 'Good' if mpg >= 25, 'Average' if 15 <= mpg < 25, else 'Poor'
df['efficiency'] = df['mpg'].apply(lambda x: 'Good' if x >= 25 else ('Average' if 15 <= x < 25 else 'Poor'))
print(df[['name', 'mpg', 'efficiency']])
```

```python
# Q6.3 Drop the name column and store the result in a new DataFrame called df_numeric.
print(df_numeric := df.drop(columns=['name']))
```

```python
# Q6.4 Add a new column hp_per_weight by dividing horsepower by weight for every row.
df_numeric['hp_per_weight'] = df_numeric['horsepower'] / df_numeric['weight']
print(df_numeric['hp_per_weight'])
```

```python
# Q7.1 Group by origin and find the average mpg for each origin.
print(df.groupby('origin')['mpg'].mean())
```

```python
# Q7.2 Group by cylinders and find the average mpg and the maximum horsepower for each group using .agg().
print(df.groupby('cylinders').agg({'mpg': 'mean', 'horsepower': 'max'}))
```

```python
# Q7.3 Group by model_year and count how many cars were made each year.
print(df.groupby('model_year').size())
```

```python
# Q7.4 Group by both origin and cylinders together, and find the average mpg for each combination.
print(df.groupby(['origin', 'cylinders'])['mpg'].mean())
```

```python
# Q8.1 Use .value_counts() to see how many cars come from each origin.
print(df['origin'].value_counts())
```

```python
# Q8.2 Find the number of unique values in cylinders and print all the unique values in model_year.
print(f"Number of unique values in cylinders: {df['cylinders'].nunique()}")
print(f"Unique values in model_year: {df['model_year'].unique()}")
```

```python
# Q8.3 Using .value_counts(), find which model_year has the highest number of cars.
print(df['model_year'].value_counts().idxmax())
```

```python
# Q9.1 Create a new column brand by extracting the first word from the name column (e.g. 'chevrolet chevelle malibu' → 'chevrolet') using .str.split().
df['brand'] = df['name'].str.split().str[0]
print(df[['name', 'brand']])
```

```python
# Q9.2 Create a new column name_upper by converting the name column to uppercase using .str.upper().
df['name_upper'] = df['name'].str.upper()
print(df[['name', 'name_upper']])
```

```python
# Q9.3 Use .str.contains('ford', case=False) to filter all cars whose name contains 'ford'.
print(df[df['name'].str.contains('ford', case=False)])
```

```python
# Q10.1 Check whether df has any duplicate rows using .duplicated(), then remove them with .drop_duplicates().
print(f"Number of duplicate rows: {df.duplicated().sum()}")
df.drop_duplicates(inplace=True)
print(f"Number of rows after removing duplicates: {df.shape[0]}")
```

```python
# Q10.2 Rename the model_year column to year and the horsepower column to hp.
df.rename(columns={'model_year': 'year', 'horsepower': 'hp'}, inplace=True)
print(df)
```

```python
# Q10.3 Rename every column name to uppercase in one line, using .rename(columns=...) with a dictionary comprehension.
df.rename(columns={col: col.upper() for col in df.columns}, inplace=True)
print(df)
```

```python
# Q11.1 Create a small lookup DataFrame mapping origin codes to full names ('usa' -> 'United States', 'japan' -> 'Japan', 'europe' -> 'Europe') and merge it with df on the origin column.
lookup_df = pd.DataFrame({
    'ORIGIN': ['usa', 'japan', 'europe'],
    'ORIGIN_FULL': ['United States', 'Japan', 'Europe']
})
merged_df = df.merge(lookup_df, on='ORIGIN', how='left')
print(merged_df[['ORIGIN', 'ORIGIN_FULL']].head())
```

```python
# Q11.2 Create two smaller DataFrames — the first 5 rows and the last 5 rows of df — and combine them back together using pd.concat().
first_five = df.head(5)
last_five = df.tail(5)
combined_df = pd.concat([first_five, last_five])
print(combined_df)
```

```python
# Q11.3 From the merged DataFrame in Q11.1, display only the name, origin, and the new full-name column for the first 5 rows.
print(merged_df[['NAME', 'ORIGIN', 'ORIGIN_FULL']].head(5))
```

```python
# Q12.1 Filter cars with mpg > 25, then save the result to a new file called efficient_cars.csv (without the index column).
efficient_cars = df[df['MPG'] > 25]
efficient_cars.to_csv('efficient_cars.csv', index=False)
print("Filtered cars saved to efficient_cars.csv")
```

```python
# Q12.2 Filter cars with cylinders == 8 and save them to a file called eight_cylinder_cars.csv.
eight_cylinder_cars = df[df['CYLINDERS'] == 8]
eight_cylinder_cars.to_csv('eight_cylinder_cars.csv', index=False)
print("Eight-cylinder cars saved to eight_cylinder_cars.csv")
```
