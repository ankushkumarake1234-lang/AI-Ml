```python
# questions no 2
plt.Figure(figsize=(7,3))
sns.countplot(data=mpg, x='origin')
plt.title("Origin Data")
plt.show()
```

```python
# Bivariate — two columns together
# 1. boxplot
# 2. scatterplot

# boxplot => does a number differ across groups?
plt.Figure(figsize=(7,3))
sns.boxplot(data=df, x='Fare')
plt.title("Fare Boxplot")
plt.show()
```

```python
# Fare split by Survived
plt.Figure(figsize=(7,3))
sns.boxplot(data=df, x='Survived', y='Fare')
plt.title("Fare Boxplot by Survived")
plt.show()
```

```python
# Age split by Survived
plt.Figure(figsize=(7,3))
sns.boxplot(data=df, x='Survived', y='Age')
plt.title("Age split by Survived")
plt.show()
```

```python
# scatterplot — do two numbers move together?
# Age against Fare
plt.Figure(figsize=(7,3))
sns.scatterplot(data=df, x='Age', y='Fare')
plt.title("Age vs Fare")
plt.show()
```

```python
# hue= — the argument that turns any plot into an ML plot
# 1. the same scatter, coloured by Survived
plt.Figure(figsize=(7,3))
sns.scatterplot(data=df, x='Age', y='Fare', hue='Survived')
plt.title("Age vs Fare")
plt.show()
```

```python
# 2. a countplot of Pclass, coloured by Survived
plt.Figure(figsize=(7,3))
sns.countplot(data=df, x='Pclass', hue='Survived')
plt.title("Pclass vs Survived")
plt.show()
```

```python
# Q3 Boxplot mpg split by origin, then split by cylinders instead. Which of the two separates the groups more clearly?
plt.Figure(figsize=(7,3))
sns.boxplot(data=mpg, x='origin', hue='cylinders')
plt.title("Origin vs Cylinders")
plt.show()
```

```python
# Q4 Scatter weight against mpg, coloured by origin. Describe the relationship in one line, and say what the colours add.

plt.Figure(figsize=(7,3))
sns.scatterplot(data=mpg, x='weight', y='mpg', hue='origin')
plt.title("Weight vs MPG")
plt.show()
```
